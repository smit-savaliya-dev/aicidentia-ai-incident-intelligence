## 1. What changed after ChatGPT

Chi-square test of independence (period x theme / domain), adjusted standardised residuals (|r| > 1.96 drives the change) and Cramer's V. Window starts at the first year of the trend window; MIT shares use labelled incidents only.

```python
for name in ("chatgpt_themes", "chatgpt_domains"):
    r = res.facts[name]
    print(name, {k: r[k] for k in ("chi2", "dof", "p_value", "cramers_v", "n_before", "n_after")})
display(res.chatgpt_domains["table"].round(3))
display(Image(filename=FIG / "fig_story_1_chatgpt_shift.png"))
```

## 2. Risk priority matrix
